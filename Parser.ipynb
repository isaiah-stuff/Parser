```python
# Parser.ipynb

# Global variables
char_class = -1
next_char = ''
next_token = -1
lexeme = []
index = 0

# Character classes (constants)
LETTER = 0
DIGIT = 1
UNKNOWN = 99

# Token codes (more constants)
INT_LIT = 10
IDENT = 11
ASSIGN_OP = 20
ADD_OP = 21
SUB_OP = 22
MULT_OP = 23
DIV_OP = 24
LEFT_PAREN = 25
RIGHT_PAREN = 26
EOF = -1

# expr function
def expr():
    print('Enter <expr>')
    
    term()
    
    global next_token
    while next_token == ADD_OP or next_token == SUB_OP:
        next_token = lex()
        term()
        
    print('Exit <expr>')

# term function
def term():
    print('Enter <term>')
  
    factor()

    global next_token
    while (next_token == MULT_OP) or (next_token == DIV_OP):
        next_token = lex()
        factor()
        
    print('Exit <term>')

# factor function
def factor():
    print('Enter <factor>')

    global next_token
    if next_token == IDENT or next_token == INT_LIT:
        next_token = lex()
    elif next_token == LEFT_PAREN:
        next_token = lex()
        expr()
        if next_token == RIGHT_PAREN:
            next_token = lex()
        else:
            error("Right parenthesis ')' expected")
    else:
        error('Invalid RHS')
        
    print('Exit <factor>')

# error function
def error(msg=''):
    print('Syntax Error:', msg)

# lookup function
def lookup(ch):
    global next_token
    if ch == '(':
        next_token = LEFT_PAREN
    elif ch == ')':
        next_token = RIGHT_PAREN
    elif ch == '+':
        next_token = ADD_OP
    elif ch == '-':
        next_token = SUB_OP
    elif ch == '*':
        next_token = MULT_OP
    elif ch == '/':
        next_token = DIV_OP
    else:
        next_token = EOF
    return next_token

# add_char function
def add_char():
    global next_char
    global lexeme
    lexeme.append(next_char)

# get_char function
def get_char():
    global next_char
    global char_class
    global index
    global line
    if index < len(line):
        next_char = line[index]
        index += 1
        if next_char.isalpha():
            char_class = LETTER
        elif next_char.isdigit():
            char_class = DIGIT
        else:
            char_class = UNKNOWN
    else:
        char_class = EOF

# get_non_blank function
def get_non_blank():
    while next_char == ' ':
        get_char()

# lex function
def lex():
    global next_char
    global char_class
    global next_token
    global lexeme
    lexeme = []
    get_non_blank()
    if char_class == LETTER:
        add_char()
        get_char()
        while char_class == LETTER or char_class == DIGIT:
            add_char()
            get_char()
        next_token = IDENT
    elif char_class == DIGIT:
        add_char()
        get_char()
        while char_class == DIGIT:
            add_char()
            get_char()
        next_token = INT_LIT
    elif char_class == UNKNOWN:
        lookup(next_char)
        lexeme.append(next_char)       
        get_char()
    elif char_class == EOF:
        next_token = EOF
        lexeme = ['E','O','F']

    lexeme_str = ''.join(lexeme)
    print(f"Next token is {next_token}, next lexeme is '{lexeme_str}'")

    return next_token

# Read the expression from file 'front.in'
f = open('front.in', mode='r')
line = f.readlines()[0]
print(f'Line: {line}')

# Call the functions to get started
get_char()
next_token = lex()
expr()
```
